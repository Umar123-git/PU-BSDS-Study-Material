## Q3–AverageHeartRatePerDoctor

```python
data={}
for entry in structured_logs:
    doctor = entry["doctor"]
    hr = int(entry["hr"])

    if doctor not in data:
        data[doctor]=[]
    data[doctor].append(hr)


avg={}
for doctor, hr in data.items():
    avg[doctor] = sum(hr)/len(hr)

for doctor, avg in avg.items():
    print(f"doctor {doctor}: Average heartrate = {avg:.2f}")
```

## Q13–DoctorSummary

```python

ds={}
for v in structured_logs:
    if v["doctor"]!=[]:
        hr = int(v["hr"])
        gl = int(v["glucose"])
        viol = str(v["doctor"])
        viol = viol.strip("[]",).strip("'")
        
        if viol not in ds:
            ds[viol] = {"count": 0, "hr": 0, "gl": 0}
        
        ds[viol]["count"] += 1
        ds[viol]["hr"] += hr
        ds[viol]["gl"] += gl
    
for d in ds.items():
    print(d)
```
